## <span style="color: #0055a5;">1. Introduction: What is LaTeX?</span>

Unlike Microsoft Word (which is WYSIWYG—*What You See Is What You Get*), **LaTeX** operates more like programming: you use markup language to format your text, then compile the source code to produce a publication-quality PDF.

* <span style="color: #2e7d32;">**Why use LaTeX?**</span> Easily formatted and beautiful scientific documents, created by scientists for scientists, with powerful packages for equations, tables, and bibliographies.
* <span style="color: #2e7d32;">**Who uses it?**</span> Students, researchers, and instructors across STEM, Economics, Linguistics, Business, Political Science, and more.

## <span style="color: #0055a5;">3. Formatting Text & Lists</span>

LaTeX commands always begin with a backslash (`\`).

* **Comments:** Anything following a `<span style="color: #d32f2f;">%</span>` symbol is ignored during compilation.
* **Bold text:** `\textbf{yourtext}`
* **Italic text:** `\textit{yourtext}`

```python
% Bulleted List
\begin{itemize}
    \item{list item 1}
    \item{list item 2}
    \item{list item 3}
\end{itemize}

% Numbered List
\begin{enumerate}
    \item{list item 1}
    \item{list item 2}
    \item{list item 3}
\end{enumerate}
```

## <span style="color: #0055a5;">4. Mathematical Expressions</span>

LaTeX is the gold standard for rendering scientific formulas. Use the `amsmath` package in your preamble (`\usepackage{amsmath}`).

* **Inline Math:** Wrapped in `\( ... \)` (e.g., $E = mc^2$).
* **Display Math (Numbered):** Wrapped in `\begin{equation} ... \end{equation}`.

## <span style="color: #0055a5;">5. Document Structure & Cross-References</span>

A standard research document skeleton requires defining a document class, metadata, and sections.

```python
\documentclass{article}
\usepackage{amsmath}

\title{My Research Paper}
\author{A. Student}
\date{\today}

\begin{document}

\maketitle

\begin{abstract}
This is the abstract content summarizing our findings.
\end{abstract}

\section{Introduction}
\label{sec:intro}
In Section \ref{sec:method}, we detail our experimental approach.

\section{Method}
\label{sec:method}
We investigate LaTeX markup and automated numbering.

\end{document}
```

## <span style="color: #0055a5;">7. Citation Basics (BibTeX & JabRef)</span>

* **Separation of Concerns:** Reference metadata is stored in a separate plain-text database file named `references.bib`.
* **Co-Presentation Note:** Covered in depth during the companion workshop co-presented by **Grace O'Hanlon & Justin Fuhr** on JabRef and Better BibTeX.

### Core Citation Commands in LaTeX:
* `\cite{key}` — Inserts an inline citation pointing to an ID in your `.bib` file.
* `\bibliography{references}` — Points to your bibliography database file.
* `\bibliographystyle{plain}` — Sets the formatting style for citations and the reference list.
