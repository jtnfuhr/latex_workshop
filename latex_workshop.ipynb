## 1. Introduction: What is LaTeX?

Unlike Microsoft Word (which is WYSIWYG—*What You See Is What You Get*), **LaTeX** operates more like programming: you use markup language to format your text, then compile the source code to produce a publication-quality PDF[cite: 1, 3].

* **Why use LaTeX?** Easily formatted and beautiful scientific documents, created by scientists for scientists, with powerful packages for equations, tables, and bibliographies[cite: 1, 3].
* **Who uses it?** Students, researchers, and instructors across STEM, Economics, Linguistics, Business, Political Science, and more[cite: 1, 3].

## 3. Formatting Text & Lists

LaTeX commands always begin with a backslash (`\`)[cite: 1, 3].

* **Comments:** Anything following a `%` symbol is ignored during compilation[cite: 1, 3].
* **Bold text:** `\textbf{yourtext}`[cite: 1, 3]
* **Italic text:** `\textit{yourtext}`[cite: 1, 3]

```python
% Bulleted List
\begin{itemize}
    \item{list item 1}
    \item{list item 2}
    \item{list item 3}
\end{itemize}

% Numbered List
\begin{enumerate}
    \item{list item 1}
    \item{list item 2}
    \item{list item 3}
\end{enumerate}
```

## 4. Mathematical Expressions

LaTeX is the gold standard for rendering scientific formulas. Use the `amsmath` package in your preamble (`\usepackage{amsmath}`)[cite: 1, 3].

* **Inline Math:** Wrapped in `\( ... \)` (e.g., $E = mc^2$)[cite: 1, 3].
* **Display Math (Numbered):** Wrapped in `\begin{equation} ... \end{equation}`[cite: 1, 3].

## 5. Document Structure & Cross-References

A standard research document skeleton requires defining a document class, metadata, and sections[cite: 1, 3].

```python
\documentclass{article}
\usepackage{amsmath}

\title{My Research Paper}
\author{A. Student}
\date{\today}

\begin{document}

\maketitle

\begin{abstract}
This is the abstract content summarizing our findings.
\end{abstract}

\section{Introduction}
\label{sec:intro}
In Section \ref{sec:method}, we detail our experimental approach.

\section{Method}
\label{sec:method}
We investigate LaTeX markup and automated numbering.

\end{document}
```

## 7. Citation Basics (BibTeX & JabRef)

* **Separation of Concerns:** Reference metadata is stored in a separate plain-text database file named `references.bib`[cite: 3].
